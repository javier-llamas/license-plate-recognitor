```python
import pathlib

import cv2
import pillow_heif
from fast_alpr import ALPR
from PIL import Image

# You can also initialize the ALPR with custom plate detection and OCR models.
alpr = ALPR(
    detector_model="yolo-v9-s-608-license-plate-end2end",
    ocr_model="cct-xs-v1-global-model",
)
```

```python
for jpg_file in pathlib.Path('assets/jpg').glob('*.jpeg'):
    frame = cv2.imread(str(jpg_file))
    assert frame is not None
    alpr.predict(frame)
    annotated_frame = alpr.draw_predictions(frame)
    output_path = jpg_file.parent / (jpg_file.stem + '_annotated.jpg')
    cv2.imwrite(str(output_path), annotated_frame)
```

```python
# Register the HEIC opener with Pillow
pillow_heif.register_heif_opener()

# iterate over all HEIC files in the assets directory
for heic_file in pathlib.Path('assets/heic').glob('*.HEIC'):
    # Open the HEIC file
    img = Image.open(heic_file)

    # Save as JPG
    image_path = heic_file.parent.parent / "jpg" / (heic_file.stem + '.jpg')
    print(f"Converting {heic_file} to {image_path}")
    img.save(image_path, format='JPEG')
    alpr_results = alpr.predict(str(image_path))
    try:
        print(alpr_results[0].ocr.text, alpr_results[0].ocr.confidence)
    except IndexError:
        print(f"Error processing {image_path}")
```
